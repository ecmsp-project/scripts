```python
import uuid
import random
from datetime import datetime
```

```python
category_names = [
    "Clothes",
    "Electronics",
    "Books",
    "Beauty",
    "Sports",
    "Toys",
    "Food"
]

# category_dict: dict[str, uuid.UUID | None] = {category_name: None for category_name in category_names}
category_dict = {}
```

```python
def generate_categories(file, category_names):
    text = f'INSERT INTO categories(id, parent_category_id, name)\n'
    file.write(text)

    text = f'VALUES\n'
    file.write(text)

    n = len(category_names)
    for category_name in category_names[:n-1]:
        category_id = uuid.uuid4()
        category_dict[category_name] = category_id

        text = f"\t('{category_id}', NULL, '{category_name}'),\n"
        file.write(text)
    #
    
    category_name = category_names[-1]
    category_id = uuid.uuid4()
    category_dict[category_name] = category_id
    text = f"\t('{category_id}', NULL, '{category_name}');\n\n"
    file.write(text)
#
```

```python
with open("sample_data.sql", "a", encoding="utf-8") as file:
    generate_categories(file, category_names)
```

```python
def generate_variants(file):
    text = f'INSERT INTO variants(id, product_id, price, stock_quantity, image_url, additional_properties, description, created_at, updated_at)\n'
    file.write(text)
    
    text = f'VALUES\n'
    file.write(text)

    variants_list = sum(variants_dict.values(), [])
    n = len(variants_list)
    for variant in variants_list[:n-1]:
        text = f"\t('{variant.id}', '{variant.product_id}', {variant.price}, {variant.stock_quantity}, '{variant.image_url}', {variant.additional_properties}, '{variant.description}', '{datetime.now()}', '{datetime.now()}'),\n"
        file.write(text)
    #

    variant = variants_list[-1]
    text = f"\t('{variant.id}', '{variant.product_id}', {variant.price}, {variant.stock_quantity}, '{variant.image_url}', {variant.additional_properties}, '{variant.description}', '{datetime.now()}', '{datetime.now()}');\n\n"
    file.write(text)
#
```

```python
def generate_categories_with_parent_id(file, category_names_with_parent):
    text = f'INSERT INTO categories(id, parent_category_id, name)\n'
    file.write(text)

    text = f'VALUES\n'
    file.write(text)

    n = len(category_names_with_parent)
    categories = list(category_names_with_parent.items())

    for category_name, category_parent_id in categories[:n-1]:
        category_id = uuid.uuid4()
        category_dict[category_name] = category_id

        text = f"\t('{category_id}', '{category_parent_id}', '{category_name}'),\n"
        file.write(text)
    #
    
    category_name, category_parent_id = categories[-1]
    category_id = uuid.uuid4()
    category_dict[category_name] = category_id
    text = f"\t('{category_id}', '{category_parent_id}', '{category_name}');\n\n"
    file.write(text)
#
```
