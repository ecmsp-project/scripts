```python
def generate_categories(file, category_names):
    text = f'INSERT INTO categories(id, parent_category_id, name)\n'
    file.write(text)

    text = f'VALUES\n'
    file.write(text)

    n = len(category_names)
    for category_name in category_names[:n-1]:
        category_id = uuid.uuid4()
        category_dict[category_name] = category_id

        text = f"\t('{category_id}', NULL, '{category_name}'),\n"
        file.write(text)
    #
    
    category_name = category_names[-1]
    category_id = uuid.uuid4()
    category_dict[category_name] = category_id
    text = f"\t('{category_id}', NULL, '{category_name}');\n\n"
    file.write(text)
#
```

```python
with open("sample_data.sql", "a", encoding="utf-8") as file:
    generate_categories(file, category_names)
```

```python
def generate_categories_with_parent_id(file, category_names_with_parent):
    text = f'INSERT INTO categories(id, parent_category_id, name)\n'
    file.write(text)

    text = f'VALUES\n'
    file.write(text)

    n = len(category_names_with_parent)
    categories = list(category_names_with_parent.items())

    for category_name, category_parent_id in categories[:n-1]:
        category_id = uuid.uuid4()
        category_dict[category_name] = category_id

        text = f"\t('{category_id}', '{category_parent_id}', '{category_name}'),\n"
        file.write(text)
    #
    
    category_name, category_parent_id = categories[-1]
    category_id = uuid.uuid4()
    category_dict[category_name] = category_id
    text = f"\t('{category_id}', '{category_parent_id}', '{category_name}');\n\n"
    file.write(text)
#
```

```python
def generate_products(file):
    text = f'INSERT INTO products(id, category_id, name, approximate_price, delivery_price, description, info)\n'
    file.write(text)
    
    text = f'VALUES\n'
    file.write(text)

    n = len(products)
    for product in products[:n-1]:
        text = f"\t('{product.id}', '{product.category_id}', '{product.name}', {product.approximate_price}, {product.delivery_price}, '{product.description}', {product.info}),\n"
        file.write(text)
    #

    product = products[-1]    
    text = f"\t('{product.id}', '{product.category_id}', '{product.name}', {product.approximate_price}, {product.delivery_price}, '{product.description}', {product.info});\n\n"
    file.write(text)
#
```

```python
def generate_variants(file):
    text = f'INSERT INTO variants(id, product_id, price, stock_quantity, additional_properties, description, created_at, updated_at)\n'
    file.write(text)
    
    text = f'VALUES\n'
    file.write(text)

    variants_list = sum(variants_dict.values(), [])
    n = len(variants_list)
    for variant in variants_list[:n-1]:
        text = f"\t('{variant.id}', '{variant.product_id}', {variant.price}, {variant.stock_quantity}, {variant.additional_properties}, '{variant.description}', '{datetime.now()}', '{datetime.now()}'),\n"
        file.write(text)
    #

    variant = variants_list[-1]
    text = f"\t('{variant.id}', '{variant.product_id}', {variant.price}, {variant.stock_quantity}, {variant.additional_properties}, '{variant.description}', '{datetime.now()}', '{datetime.now()}');\n\n"
    file.write(text)
#
```

```python
with open("sample_data.sql", "a", encoding="utf-8") as file:
    generate_products(file)
    generate_variants(file)
```

```python
def generate_properties(file, properties_list):
    text = f'INSERT INTO properties(id, category_id, name, unit, data_type, has_default_options, role)\n'
    file.write(text)
    
    text = f'VALUES\n'
    file.write(text)

    n = len(properties_list)
    for property in properties_list[:n-1]:
        text = f"\t('{property.id}', '{property.category_id}', '{property.name}', {property.unit}, '{property.data_type}', '{property.has_default_options}', '{property.role}'),\n"
        file.write(text)
    #

    property = properties_list[-1]
    text = f"\t('{property.id}', '{property.category_id}', '{property.name}', {property.unit}, '{property.data_type}', '{property.has_default_options}', '{property.role}');\n\n"
    file.write(text)
#
```

```python
def generate_default_property_options(file, default_property_options_list):
    text = f'INSERT INTO default_property_options(id, property_id, value_text, value_decimal, value_boolean, value_date, display_text)\n'
    file.write(text)
    
    text = f'VALUES\n'
    file.write(text)

    n = len(default_property_options_list)
    for dpo in default_property_options_list[:n-1]:
        text = f"\t('{dpo.id}', '{dpo.property_id}', {dpo.value_text}, {dpo.value_decimal}, {dpo.value_boolean}, {dpo.value_date}, '{dpo.display_text}'),\n"
        file.write(text)
    #

    dpo = default_property_options_list[-1]
    text = f"\t('{dpo.id}', '{dpo.property_id}', {dpo.value_text}, {dpo.value_decimal}, {dpo.value_boolean}, {dpo.value_date}, '{dpo.display_text}');\n\n"
    file.write(text)
#
```

```python
def generate_variant_properties(file, variant_properties_list):
    text = f'INSERT INTO variant_properties(id, variant_id, property_id, value_text, value_decimal, value_boolean, value_date, display_text)\n'
    file.write(text)
    
    text = f'VALUES\n'
    file.write(text)

    n = len(variant_properties_list)
    for vp in variant_properties_list[:n-1]:
        text = f"\t('{vp.id}', '{vp.variant_id}', '{vp.property_id}', {vp.value_text}, {vp.value_decimal}, {vp.value_boolean}, {vp.value_date}, '{vp.display_text}'),\n"
        file.write(text)
    #

    vp = variant_properties_list[-1]
    text = f"\t('{vp.id}', '{vp.variant_id}', '{vp.property_id}', {vp.value_text}, {vp.value_decimal}, {vp.value_boolean}, {vp.value_date}, '{vp.display_text}');\n\n"
    file.write(text)
#
```
